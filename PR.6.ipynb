```python
import os
from datetime import datetime

class JournalManager:
    def __init__(self, filename="journal.txt"):
        self.filename = filename

    def add_entry(self):
        entry_text = input("\nEnter your journal entry:\n")
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        
        try:
            with open(self.filename, "a", encoding="utf-8") as file:
                file.write(f"[{timestamp}]\n")
                file.write(f"{entry_text}\n\n")
            print("\nEntry added successfully!")
        except PermissionError:
            print("\nError: Permission denied. Unable to write to file.")

    def view_all_entries(self):
        if not os.path.exists(self.filename) or os.path.getsize(self.filename) == 0:
            print("\nOutput (If the file does not exist):")
            print("No journal entries found. Start by adding a new entry!")
            return

        try:
            with open(self.filename, "r", encoding="utf-8") as file:
                content = file.read().strip()
                
                print("\nOutput (If the file exists):")
                print("Your Journal Entries:")
                print("---------------------")
                print(content)
        except PermissionError:
            print("\nError: Permission denied. Unable to read the file.")

    def search_entry(self):
        if not os.path.exists(self.filename) or os.path.getsize(self.filename) == 0:
            print("\nOutput:")
            print("Error: The journal file does not exist. Please add a new entry first.")
            return

        keyword = input("\nEnter a keyword or date to search: ").strip().lower()
        
        try:
            with open(self.filename, "r", encoding="utf-8") as file:
                content = file.read().strip()
                
            entries = content.split("\n\n")
            matches = [entry for entry in entries if keyword in entry.lower()]
            
            if matches:
                print("\nOutput (If a match is found):")
                print("Matching Entries:")
                print("-----------------")
                for match in matches:
                    print(match)
                    print()
            else:
                print("\nOutput (If no match is found):")
                print(f"No entries were found for the keyword: {keyword}.")
                
        except Exception as e:
            print(f"\nAn error occurred during search: {e}")

    def delete_all_entries(self):
        if not os.path.exists(self.filename):
            print("\nOutput (If the file does not exist):")
            print("No journal entries to delete.")
            return

        confirm = input("\nAre you sure you want to delete all entries? (yes/no): ").strip().lower()
        
        if confirm == 'yes':
            try:
                os.remove(self.filename)
                print("\nOutput (If the file is deleted successfully):")
                print("All journal entries have been deleted.")
            except Exception as e:
                print(f"\nError: Could not delete file: {e}")
        else:
            print("\nDeletion cancelled.")


def main():
    manager = JournalManager()
    
    while True:
       
        print("\nWelcome to Personal Journal Manager!")
        print("Please select an option:")
        print("\n1. Add a New Entry")
        print("2. View All Entries")
        print("3. Search for an Entry")
        print("4. Delete All Entries")
        print("5. Exit")
        
        choice = input("\nUser Input:\n").strip()
        
        if choice == '1':
            manager.add_entry()
        elif choice == '2':
            manager.view_all_entries()
        elif choice == '3':
            manager.search_entry()
        elif choice == '4':
            manager.delete_all_entries()
        elif choice == '5':
            print("\nOutput:")
            print("Thank you for using Personal Journal Manager. Goodbye!")
            break
        else:
            print("\nOutput:")
            print("Invalid option. Please select a valid option from the menu.")

if __name__ == "__main__":
    main()
```
